## 🏆 Step 2: Statistical Aggregation (Mean ± Std)

```python
metrics_summary = {}
print('=' * 80)
print('🎉 OFFICIAL TABLE 4 BENCHMARK SUMMARY (10-FOLD EMPIRICAL MEAN):')
print('=' * 80)

for k in ['ACC', 'F1', 'AUC', 'AUC_FS', 'AUC_FR']:
    vals = [ALL_10_FOLDS[f][k] * 100 for f in range(1, 11)]
    mean_val = np.mean(vals)
    std_val = np.std(vals)
    metrics_summary[k] = (mean_val, std_val)
    print('   • {:<10} : {:6.2f}% ± {:.2f}%'.format(k, mean_val, std_val))
print('=' * 80)
```
